```python
import pandas as pd
```

```python
df_train = pd.read_json(r".\train.json")
df_val = pd.read_json(r".\dev.json")
df_test = pd.read_json(r".\test.json")
```

```python
df_train.sample(2)
```

```python
df_val.sample(2)
```

```python
df_test.sample(2)
```

```python
df = pd.concat(objs=[df_train, df_val, df_test], axis=0)
df = df.reset_index(drop=True)
len(df)
```

```python
df.columns
```

```python
df = df[["pre_text", "table", "post_text", "id", "qa"]]
df
```

```python
df.iloc[0]["qa"]
```

```python
KEEP_KEYS = {"question", "program", "exe_ans"}

df["qa"] = df["qa"].apply(lambda qa: {k: v for k, v in qa.items() if k in KEEP_KEYS})
df
```

```python
df.to_json("FinQA_processed.json", orient="records", indent=2, force_ascii=False)
```
